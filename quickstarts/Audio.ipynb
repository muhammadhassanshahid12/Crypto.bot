```
# Project 01: Student ID & Name Analyzer

# Store student information
student_name = "   Muhammad Hassan Bilwani   "
student_id = "FA24-BSCS-0255"

# Remove extra spaces
clean_name = student_name.strip()

# Convert name
upper_name = clean_name.upper()
title_name = clean_name.title()

# Indexing
first_char = clean_name[0]
last_char = clean_name[-1]

# Slicing
first_3 = clean_name[:3]
last_3 = clean_name[-3:]

# Student ID slicing
session = student_id[:4]
department = student_id[5:9]
roll_no = student_id[-4:]

# Display results
print("===== Student ID & Name Analyzer =====")
print("Original Name:", student_name)
print("Clean Name:", clean_name)
print("Uppercase Name:", upper_name)
print("Title Case Name:", title_name)

print("\nFirst Character:", first_char)
print("Last Character:", last_char)

print("\nFirst 3 Characters:", first_3)
print("Last 3 Characters:", last_3)

print("\nStudent ID:", student_id)
print("Session:", session)
print("Department:", department)
print("Roll Number:", roll_no)
```

```
# Project 02: Email & Username Analyzer

# Store user information
full_name = "   Muhammad Hassan Bilwani   "
email = "muhammadhassan123@gmail.com"

# Remove extra spaces
clean_name = full_name.strip()

# Convert name
lower_name = clean_name.lower()
title_name = clean_name.title()

# Indexing
first_char = clean_name[0]
third_char = clean_name[2]
last_char = clean_name[-1]

# Extract username and domain
at_position = email.index("@")

username = email[:at_position]
domain = email[at_position + 1:]

# Display results
print("===== Email & Username Analyzer =====")
print("Original Name:", full_name)
print("Clean Name:", clean_name)

print("\nLowercase Name:", lower_name)
print("Title Case Name:", title_name)

print("\nFirst Character:", first_char)
print("Third Character:", third_char)
print("Last Character:", last_char)

print("\nEmail Address:", email)
print("Username:", username)
print("Domain Name:", domain)
```

```
# Project 03: Product Code Analyzer

# Store product information
product_name = "   Wireless Mouse   "
product_code = "WMXK20240078"

# Remove extra spaces
clean_product = product_name.strip()

# Convert product name
upper_product = clean_product.upper()
lower_product = clean_product.lower()

# Indexing
first_char = clean_product[0]
last_char = clean_product[-1]

# Slicing
first_4_code = product_code[:4]
last_4_code = product_code[-4:]

# Display results
print("===== Product Code Analyzer =====")
print("Original Product Name:", product_name)
print("Clean Product Name:", clean_product)

print("\nUppercase:", upper_product)
print("Lowercase:", lower_product)

print("\nFirst Character:", first_char)
print("Last Character:", last_char)

print("\nProduct Code:", product_code)
print("First 4 Characters:", first_4_code)
print("Last 4 Characters:", last_4_code)
```
